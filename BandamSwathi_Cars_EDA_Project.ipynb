```python
# Fill missing values
df["Number of Doors"] = df["Number of Doors"].fillna(df["Number of Doors"].mode()[0])
df["Engine HP"] = df["Engine HP"].fillna(df["Engine HP"].median())
df["Engine Cylinders"] = df["Engine Cylinders"].fillna(df["Engine Cylinders"].median())
df["Market Category"] = df["Market Category"].fillna("Not Specified")
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
# Remove 3 rows with missing fuel type
df = df.dropna(subset=["Engine Fuel Type"])
```

```python
df.isnull().sum()
```

```python
# Remove duplicate rows
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
# Rename columns so they are easy to type (lowercase, no spaces)
df.columns = df.columns.str.lower().str.replace(" ", "_")
df.info()
```

```python
df["transmission_type"].value_counts()
```

```python
# Remove rows with UNKNOWN transmission (only 12 rows)
df = df[df["transmission_type"] != "UNKNOWN"]
```

```python
# Create a car age column (dataset ends at year 2017)
df["car_age"] = 2017 - df["year"]
```

```python
# Chart 1: Histogram - price distribution
plt.figure(figsize=(8, 5))
sns.histplot(df["msrp"])
plt.title("Distribution of Car Prices")
plt.xlabel("Price (MSRP)")
plt.ylabel("Number of cars")
plt.show()
```

```python
# Chart 2: Histogram - price distribution(bins=50 specifies that the entire range of car prices should be split into 50 equal-sized intervals (or "buckets").)
plt.figure(figsize=(8, 5))
sns.histplot(df["msrp"], bins=50)
plt.title("Distribution of Car Prices")
plt.xlabel("Price (MSRP)")
plt.ylabel("Number of cars")
plt.show()
```

```python
# Chart 3: Bar chart - top 10 most common brands
plt.figure(figsize=(9, 5))
df["make"].value_counts().head(10).plot(kind="bar")
plt.title("Top 10 Most Common Brands")
plt.xlabel("Brand")
plt.ylabel("Number of cars")
plt.show()
```

```python
# Chart 4: Count plot - transmission type
plt.figure(figsize=(8, 5))
sns.countplot(x="transmission_type", data=df)
plt.title("Cars by Transmission Type")
plt.xlabel("Transmission type")
plt.ylabel("Number of cars")
plt.show()
```

```python
df.info()
```

```python
# Chart 5: Box plot - price by transmission type (cars under 100000 for readability)
plt.figure(figsize=(8, 5))
sns.boxplot(x="transmission_type", y="msrp", data=df[df["msrp"] < 100000])
plt.title("Price by Transmission Type (cars under 100,000)")
plt.xlabel("Transmission type")
plt.ylabel("Price (MSRP)")
plt.show()
```

```python

# Apply log transformation
df['msrp'] = np.log1p(df['msrp'])
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x="transmission_type", y="msrp", data=df[df["msrp"] < 100000])
plt.title("Price by Transmission Type (cars under 100,000)")
plt.xlabel("Transmission type")
plt.ylabel("Price (MSRP)")
plt.show()
```
